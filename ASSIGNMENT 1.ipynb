```python
#LIST FUNCTIONS
```

```python
#CREATING A LIST
x=[1,2,3]
print(x)
```

```python
#ACCESSING LIST ITEMS
a=['x','y','z']
print(a[1])
```

```python
#CHANGING LIST ITEMS
a=['x','y','z']
a[1]='w'
print(a[1])
print(a)
```

```python
#ADDING ITEMS TO LIST
a=['x','y','z']
a.append('w')
print(a)
```

```python
#ADDING LIST TO LIST
x=[1,2,3]
a=['x','y','z']
a.extend(x)
print(a)
```

```python
#TUPLE FUNCTIONS
```

```python
#CREATING A TUPLE
t=('a','b','c')
print(t)
```

```python
#ACCESING TUPLE ITEMS
print(t[2])
```

```python
#CONVERTING TUPLE INTO LIST
r=list(t)
print(r)
```

```python
#ADDING ITEM TO TUPLE
t=('a','b','c')
r=list(t)
r.insert(0,'n')
t=tuple(r)
print(t)
```

```python
#REMOVING ITEMS FROM TUPLE
t=('a','b','c')
r=list(t)
r.remove('a')
t=tuple(r)
print(t)
```

```python
#JOINING TWO TUPLES
t=('a','b','c')
r=(1,2,3)
s=t+r
print(s)
```

```python
#DICTIONARY FUNCTIONS
```

```python
#CREATING A DICTIONARY
d={'Name': 'Akshat',
   'Age': 16, 
   'DOB': '27.10.2009'}
print(d)
```

```python
#ACCESSING DICTIONARY ITEMS
d={'Name': 'Akshat',
   'Age': 16, 
   'DOB': '27.10.2009'}
print(d['Age'])
```

```python
#GETTING KEYS OF DICTIONARY
d={'Name': 'Akshat',
   'Age': 16, 
   'DOB': '27.10.2009'}
print(d.keys())
```

```python
#CHANGING VALUES
d={'Name': 'Akshat',
   'Age': 16, 
   'DOB': '27.10.2009'}
d['Age']=17
print(d)
```

```python
#ADDING ITEMS TO DICTIONARY
d={'Name': 'Akshat',
   'Age': 16, 
   'DOB': '27.10.2009'}
d['Year']='F.Y.'
print(d)
```
